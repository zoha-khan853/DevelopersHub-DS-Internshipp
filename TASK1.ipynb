#TASK1: Exploring and Visulaizing the Iris Dataset. This notebook explores the Iris dataset using pandas, matplotlib, and seaborn.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
print("Librraies Imported successfully")
```

```python
df = sns.load_dataset('iris')
print("Dataset Loaded!")
```

```python
print("shape:", df.shape)
print("\nColumns:" , df.columns.tolist())
df.head()
```

```python
print(df.describe())
print("\nMissing values:")
print(df.isnull().sum())
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x='sepal_length', y='petal_length', hue='species')
plt.title('Sepal Length vs Petal Length')
plt.show()
```

```python
df.hist(figsize=(10,6), bins=20, color='steelblue', edgecolor='black')
plt.suptitle('Feature Distributions')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.boxplot(data=df, x='species', y='petal_length')
plt.title('Petal Length by Species')
plt.show()
```
